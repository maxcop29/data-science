```python
df['order_date']=pd.to_datetime(df['order_date'],format='mixed')
```

```python
df['country'].unique()
```

```python

country_mapping = {
    'tunisia': 'Tunisia',
    'USA': 'United States',
    'U.S.A': 'United States'
}

df['country'] = df['country'].replace(country_mapping)


df['country'] = df['country'].str.title()


print(df['country'].unique())
```

```python
df['category'].value_counts()
```

```python
df['category'] = df['category'].str.title()
```

```python
df['quantity']=df['quantity'].abs()
```

```python
df['payment_method'].value_counts()
```

```python
df['payment_method']=df['payment_method'].str.title()
```

```python
df['tax']=df['tax'].abs()
```

```python
df['currency'].value_counts()
```

```python
df['total_amount']=df['total_amount'].abs()
```

```python
df['status'].unique()
```

```python
df['status'] = df['status'].str.title()

df['status'] = df['status'].replace({'Complete': 'Completed'})
```

```python
df.isna().sum()
```

```python
df['sales_rep'] = df['sales_rep'].fillna('Unknown')
df['customer_email'] = df['customer_email'].fillna('Unknown')
```

```python
df['discount'] = df['discount'].fillna(0.0)
df['shipping_cost'] = df['shipping_cost'].fillna(0.0)
```

```python
df = df.dropna(subset=['order_date'])
```
