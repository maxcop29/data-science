```python

import time
import random
import requests
from bs4 import BeautifulSoup
import pandas as pd

HEADERS = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                  '(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

session = requests.Session()
session.headers.update(HEADERS)

df_list = []
consecutive_empty = 0
MAX_CONSECUTIVE_EMPTY = 3   # stop early if we hit several blank/blocked pages in a row

for page in range(1, 500):
    url = f'https://www.ambitionbox.com/list-of-companies?page={page}'

    try:
        resp = session.get(url, timeout=15)
        resp.raise_for_status()
    except requests.exceptions.RequestException as e:
        print(f"[page {page}] request failed: {e} — skipping")
        consecutive_empty += 1
        if consecutive_empty >= MAX_CONSECUTIVE_EMPTY:
            print("Too many consecutive failures, stopping.")
            break
        time.sleep(5)
        continue

    soup = BeautifulSoup(resp.text, 'lxml')
    company = soup.find_all('div', class_='companyCardWrapper')

    if not company:
        print(f"[page {page}] no company cards found (blocked, CAPTCHA, or end of pages)")
        consecutive_empty += 1
        if consecutive_empty >= MAX_CONSECUTIVE_EMPTY:
            print("Too many consecutive empty pages, stopping.")
            break
        time.sleep(5)
        continue

    consecutive_empty = 0  # reset on a successful page

    name, rating, reviews, jobs, industry, location = [], [], [], [], [], []

    for i in company:
        # 1. Name — guard against missing h2
        name_el = i.find('h2')
        name.append(name_el.text.strip() if name_el else None)

        # 2. Rating
        rating_el = i.find('span', class_='rating_text')
        rating.append(rating_el.text.strip() if rating_el else None)

        # 3. Reviews — match on href containing 'reviews', not just first ActionWrapper link
        rev_el = i.find('a', href=lambda h: h and 'reviews' in h.lower())
        reviews.append(rev_el.text.strip() if rev_el else None)

        # 4. Jobs
        job_el = i.find('a', href=lambda h: h and '-jobs-cmp' in h)
        jobs.append(job_el.text.strip() if job_el else None)

        # 5. Industry / location
        meta_el = i.find('span', class_='companyCardWrapper__interLinking')
        if meta_el and meta_el.text:
            meta_parts = [p.strip() for p in meta_el.text.split('|')]
            industry.append(meta_parts[0] if len(meta_parts) > 0 else None)
            location.append(meta_parts[1] if len(meta_parts) > 1 else None)
        else:
            industry.append(None)
            location.append(None)

    df = pd.DataFrame({
        'name': name,
        'rating': rating,
        'reviews': reviews,
        'jobs': jobs,
        'industry': industry,
        'location': location
    })
    df_list.append(df)
    print(f"[page {page}] scraped {len(df)} companies")

    # be polite — randomized delay to reduce block risk
    time.sleep(random.uniform(1.5, 3.0))

final = pd.concat(df_list, ignore_index=True) if df_list else pd.DataFrame()
final.to_csv('ambitionbox_companies.csv', index=False)
print(f"Done. Total rows: {len(final)}")
```
