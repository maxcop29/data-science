```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
```

```python
df.isna().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.info()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), format='%B %d, %Y', errors='coerce')
```

```python
df.head()
```

```python

df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.strftime('%B')
```

```python
text_cols = ['type', 'title', 'director', 'cast', 'country', 'rating', 'listed_in']
for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
df[['duration_value', 'duration_unit']] = df['duration'].str.extract(r'(\d+)\s*(\w+)')
df['duration_value'] = pd.to_numeric(df['duration_value'])
```

```python
# --- 1. Average Movie Duration & Release Year Trend ---
avg_movie_len = df[df['type'] == 'Movie']['duration_value'].mean()
print(f"Average Movie Duration: {avg_movie_len:.2f} minutes")

# Relationship between release year and duration (Line plot of averages per year)
plt.figure(figsize=(12, 5))
movie_trends = df[df['type'] == 'Movie'].groupby('release_year')['duration_value'].mean().reset_index()
sns.lineplot(data=movie_trends[movie_trends['release_year'] > 2000], x='release_year', y='duration_value', color='purple')
plt.title('Average Movie Duration by Release Year (Since 2000)')
plt.ylabel('Average Duration (minutes)')
plt.show()

# --- 2. Average TV Show Seasons ---
avg_tv_seasons = df[df['type'] == 'TV Show']['duration_value'].mean()
print(f"Average TV Show Duration: {avg_tv_seasons:.2f} Seasons")
```
